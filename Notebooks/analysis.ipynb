```python
from pathlib import Path

import pandas as pd

root = Path.cwd().resolve()
data_dir = (root / ".." / "phydata_capstone data").resolve()

gdp_path = data_dir / "API_NY.GDP.PCAP.CD_DS2_en_csv_v2_464091" / "API_NY.GDP.PCAP.CD_DS2_en_csv_v2_464091.csv"
agri_gdp_path = data_dir / "API_NV.AGR.TOTL.ZS_DS2_en_csv_v2_458715" / "API_NV.AGR.TOTL.ZS_DS2_en_csv_v2_458715.csv"
agri_emp_path = data_dir / "API_SL.AGR.EMPL.ZS_DS2_en_csv_v2_439956" / "API_SL.AGR.EMPL.ZS_DS2_en_csv_v2_439956.csv"

gdp = pd.read_csv(gdp_path, skiprows=4)
agri_gdp = pd.read_csv(agri_gdp_path, skiprows=4)
agri_emp = pd.read_csv(agri_emp_path, skiprows=4)
```

```python
def reshape_world_bank(df, value_name):
    df = df.drop(columns=["Indicator Name", "Indicator Code"], errors="ignore")

    df = df.melt(
        id_vars=["Country Name", "Country Code"],
        var_name="Year",
        value_name=value_name
    )

    df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

    return df
```

```python
gdp = reshape_world_bank(gdp, "GDP_per_capita")
agri_gdp = reshape_world_bank(agri_gdp, "Agriculture_GDP")
agri_emp = reshape_world_bank(agri_emp, "Agriculture_Employment")
```

```python
df = gdp.merge(
    agri_gdp,
    on=["Country Name", "Country Code", "Year"],
    how="inner"
)

df = df.merge(
    agri_emp,
    on=["Country Name", "Country Code", "Year"],
    how="inner"
)
```

```python
df = df.dropna(
    subset=[
        "GDP_per_capita",
        "Agriculture_GDP",
        "Agriculture_Employment"
    ]
)
```
